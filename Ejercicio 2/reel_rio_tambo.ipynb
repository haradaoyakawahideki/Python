## Pasos 2 y 3: el guion revisado

Cada escena tiene una `frase` (se dice, con los números en palabras), una `cifra` y una `etiqueta` (se ven en pantalla). Todas las cifras vienen del README. En la revisión se quitó «Oe, causa» al inicio.

```python
import json

guion = json.load(open("guion.json"))
for i, e in enumerate(guion["escenas"], 1):
    print(f"{i}. [{e['cifra'] or '—'}] {e['etiqueta']}\n   {e['frase']}\n")
```

## Paso 4: la voz

`voz.py` clona la voz de `Tarea Diplomados.m4a` (19,8 s) con `Qwen/Qwen3-TTS-12Hz-0.6B-Base` en CPU (en el GPU de la M1 falla el codificador de audio). Whisper vuelve a escuchar cada escena y la rechaza si:

- alguna cifra del guion no se oye tal cual (p. ej. «ciento veintitrés» salió como 233, y «mil seiscientas nueve» como 1 109);
- no se oye el final de la frase;
- entiende mal más del 25 % de las palabras.

Cada escena se corta 0,3 s después de su última palabra, porque a veces el modelo sigue generando ruido. Tarda varios minutos; si se interrumpe, retoma desde las escenas ya guardadas.

```python
!caffeinate -i .venv/bin/python voz.py
```
